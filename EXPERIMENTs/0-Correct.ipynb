```python
from schemes import fdb
```

```python
from collections import namedtuple
```

```python
SQL_Query = namedtuple("SQL_Query", ["Select", "Where", "Value"])
```

```python
test_folder = ["../data/sf0.001"]
```

```python
ct_fdb = fdb.Client()
```

```python
ct_fdb.load_tables(test_folder[0], tb_list)
```
